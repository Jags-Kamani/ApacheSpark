Question:

Write a solution to find managers with at least five direct reports.
Return the result table in any order.

**Sample Input:**

Employee table:



```
    id   name department  managerId
0  101   John          A        NaN
1  102    Dan          A      101.0
2  103  James          A      101.0
3  104    Amy          A      101.0
4  105   Anne          A      101.0
5  106    Ron          B      101.0
```



**Sample Output**



```
   name
0  John
```



**Your answer here**


````python
#Scala-Spark

import org.apache.spark.sql.SparkSession
import org.apache.spark.sql.functions._

val spark = SparkSession.builder.appName("Aggregation Example").master[*].getOrCreate()

import spark.implicits._

val df = Seq(
    (101, John, A, NaN),
    (102, Dan, A, 101.0),
    (103, James, A, 101.0),
    (104, Amy, A, 101.0),
    (105, Anne, A, 101.0),
    (106, Ron, B, 101.0)
).toDf("id", "name", "department", "managerId")

val result_df= df.groupBy("managerId")
                 .agg(count(*).alias("direct_reports"))
                 .filter("direct_reports" > 5)
                 .join(df, df("id") === col("managerId"), left)
                 .select("name")

result_df.show()

````

**Expected Output**



```
  name
0  Bob
```
