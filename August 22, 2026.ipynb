```python
list = ["A~B", "C~D"]
print()
print("====lisstr====")
print(list)

rddst = sc.parallelize(list)
print()
print("====rddstr====")
print(rddst.collect())

flatmapsplit = rddst.flatMap(lambda x : x.split("~"))
print()
print("====flatMap====")
print(flatmapsplit.collect())

mapsplit  = rddst.map(  lambda  x : x.split("~"))
print()
print("====mapsplit=====")
print(mapsplit.collect())
```
