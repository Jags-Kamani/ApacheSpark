# **1. Filter Rows**

Question 1: From the dataset below, filter all rows where City is "NY" and Discount_Applied is "Yes".

````python
import pandas as pd
data = {
    "Customer_ID": [1, 2, 3, 4],
    "Age": [24, 45, 31, 28],
    "Gender": ["Male", "Female", "Male", "Female"],
    "Purchase_Amount_USD": [253, 364, 540, 120],
    "Discount_Applied": ["Yes", "No", "Yes", "No"],
    "City": ["NY", "SF", "NY", "Chicago"]
}
df = pd.DataFrame(data)
````

**Your Answer Below**

````python
#Scala-Spark
import org.apache.spark.sql.SparkSession
import org.apache.spark.sql.functions._

val spark = SparkSession.builder().appName("Filter Example").master(Local[*]).getOrCreate()

import spark.implicits._

val df = Seq(
    (1, 24, "Male", 253, "Yes", "NY"),
    (2, 45, "Female", 364, "No", "SF"),
    (3, 31, "Male", 540, "Yes", "NY"),
    (4, 28, "Female", 120, "No", "Chicago")
).toDf("Customer_ID: long, Age: long, Gender: string, Purchase_Amount_USD: long, Discount_Applied: string, City: string")

val filtered_df = df.filter($"City" == "NY" && $"Discount_Applied" == "Yes")

filtered_df.show()
````

**Expected_Output**

```
   Customer_ID  Age Gender  Purchase_Amount_USD Discount_Applied City
0            1   24   Male                  253              Yes   NY
2            3   31   Male                  540              Yes   NY
```



**Solution:    ⚠️ Please do not open the Solution before solving**


````python
# @title
df = df[(df["City"] == "NY") & (df["Discount_Applied"] == "Yes")]
````
